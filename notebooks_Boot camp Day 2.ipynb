```python
#play and match list also randomly select the score for the player, summation of the score, more tha average score palyers, max scored palyers, variability in score

# Player names
players = np.array(["Virat", "Rohit", "Dhoni", "Rahul", "Gill"])

# Randomly generate scores between 0 and 100
scores = np.random.randint(0, 101, size=5)

print("Players:", players)
print("Scores:", scores)

# 1. Sum of all scores
total = np.sum(scores)
print("Total Score:", total)

# 2. Average score
average = np.mean(scores)
print("Average Score:", average)

# 3. Players scoring more than average
print("Players scoring more than average:")

for i in range(len(players)):
    if scores[i] > average:
        print(players[i], scores[i])

# 4. Maximum scored player
max_score = np.max(scores)

print("Maximum Score:", max_score)

for i in range(len(players)):
    if scores[i] == max_score:
        print("Highest Scorer:", players[i])

# 5. Variability in scores
variability = np.std(scores)
print("Standard Deviation:", variability)
```

```python
import pandas as pd

# Create a Pandas Series
marks = pd.Series([85, 72, 90, 65, 78])

print("Series:")
print(marks)

# Create a DataFrame
data = {
    "Name": ["Virat", "Rohit", "Dhoni", "Rahul", "Gill"],
    "Age": [35, 37, 42, 32, 25],
    "Score": [85, 72, 90, 65, 78]
}

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)

# Access a column
print("\nScores:")
print(df["Score"])

# Access a row by index
print("\nFirst row:")
print(df.iloc[0])

# Access a particular value
print("\nFirst player's score:")
print(df.iloc[0]["Score"])

# Filter players with score > 75
print("\nPlayers scoring more than 75:")
print(df[df["Score"] > 75])

# Add a new column
df["Passed"] = df["Score"] >= 50

print("\nUpdated DataFrame:")
print(df)

# Total score
print("\nTotal Score:", df["Score"].sum())

# Average score
print("Average Score:", df["Score"].mean())

# Maximum score
print("Maximum Score:", df["Score"].max())

# Minimum score
print("Minimum Score:", df["Score"].min())

# Basic information
print("\nShape:", df.shape)
print("Columns:", df.columns)
print("Data Types:")
print(df.dtypes)
```

```python
#Student Marks
import matplotlib.pyplot as plt

# Dataset
data = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 67, 92, 74]
})

# Plot
plt.plot(data["Student"], data["Marks"], marker="o")

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid()

plt.show()
```

```python
# Generate wave data
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Store in DataFrame
data = pd.DataFrame({
    "X": x,
    "Wave": y
})

# Plot wave
plt.plot(data["X"], data["Wave"])

plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 1. Prepare data
x = np.linspace(0, 2 * np.pi, 200)
df = pd.DataFrame({"x": x, "y": np.sin(x)})

# 2. Set up figure
fig, ax = plt.subplots()
line, = ax.plot([], [], lw=2)
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.5, 1.5)

# 3. Update function
def update(frame):
    y_shifted = np.sin(df["x"] + frame / 10.0)
    line.set_data(df["x"], y_shifted)
    return line,

ani = FuncAnimation(fig, update, frames=60, interval=40, blit=True)

# Prevent Matplotlib from displaying the extra blank/static first frame
plt.close(fig)

# 4. Render directly in notebook cell
HTML(ani.to_jshtml())
```
