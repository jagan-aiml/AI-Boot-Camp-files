```python
# ============================================================
# IMAGE OBJECT DETECTION USING YOLO
# ============================================================

# 1. Install YOLO
!pip install ultralytics -q

# 2. Import libraries
from ultralytics import YOLO
from google.colab import files
from IPython.display import display
from PIL import Image
import cv2


# 3. UPLOAD IMAGE

uploaded = files.upload()

# Get uploaded image filename
image_path = list(uploaded.keys())[0]

print("Image:", image_path)


# 4. READ AND DISPLAY ORIGINAL IMAGE

img = cv2.imread(image_path)

print("Image Type:", type(img))
print("Image Shape:", img.shape)

print("\nOriginal Image:")
display(Image.open(image_path))


# 5. LOAD PRETRAINED YOLO MODEL

model = YOLO("yolo11n.pt")

print("\nYOLO Model Loaded Successfully")

# 6. DETECT OBJECTS

results = model(image_path)

print("\nObject Detection Completed")


# 7. DISPLAY DETECTION RESULTS

for result in results:

    # Get detected objects
    boxes = result.boxes

    print("\nNumber of Objects Detected:", len(boxes))

    # Print each detected object
    for box in boxes:

        class_id = int(box.cls[0])
        confidence = float(box.conf[0])

        class_name = model.names[class_id]

        print(
            "Object:",
            class_name,
            "| Confidence:",
            round(confidence * 100, 2),
            "%"
        )

    # Generate image with bounding boxes
    detected_image = result.plot()

    print("\nDetected Image:")
    cv2_imshow(detected_image)


# 8. SAVE DETECTED IMAGE

output_path = "/content/detected_image.jpg"

cv2.imwrite(output_path, detected_image)

print("\nDetected image saved at:")
print(output_path)


# 9. DISPLAY SAVED IMAGE

saved_image = cv2.imread(output_path)

print("\nSaved Detection Image:")
cv2_imshow(saved_image)
```

```python
# ============================================================
# FACE MESH DETECTION USING MEDIAPIPE
# ============================================================

# 1. INSTALL MEDIAPIPE
!pip install mediapipe -q

# 2. IMPORT LIBRARIES
import cv2
import mediapipe as mp
from google.colab import files
from google.colab.patches import cv2_imshow


# ============================================================
# 3. UPLOAD IMAGE
# ============================================================

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

print("Uploaded Image:", image_path)


# ============================================================
# 4. READ IMAGE
# ============================================================

img = cv2.imread(image_path)

print("Image Type:", type(img))
print("Image Shape:", img.shape)

print("\nOriginal Image:")
cv2_imshow(img)


# ============================================================
# 5. INITIALIZE MEDIAPIPE FACE MESH
# ============================================================

mp_face_mesh = mp.solutions.face_mesh
mp_drawing = mp.solutions.drawing_utils

face_mesh = mp_face_mesh.FaceMesh(
    static_image_mode=True,
    max_num_faces=5,
    min_detection_confidence=0.5
)


# ============================================================
# 6. CONVERT BGR TO RGB
# ============================================================

rgb_image = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2RGB
)


# ============================================================
# 7. FACE MESH DETECTION
# ============================================================

results = face_mesh.process(rgb_image)


# ============================================================
# 8. DRAW FACE LANDMARKS
# ============================================================

if results.multi_face_landmarks:

    print(
        "Number of Faces Detected:",
        len(results.multi_face_landmarks)
    )

    for face_landmarks in results.multi_face_landmarks:

        mp_drawing.draw_landmarks(
            image=img,
            landmark_list=face_landmarks,
            connections=mp_face_mesh.FACEMESH_TESSELATION
        )

else:

    print("No face detected")


# ============================================================
# 9. DISPLAY FACE MESH
# ============================================================

print("\nFace Mesh Result:")

cv2_imshow(img)


# ============================================================
# 10. SAVE RESULT
# ============================================================

output_path = "/content/face_mesh_result.jpg"

cv2.imwrite(
    output_path,
    img
)

print("\nFace mesh image saved at:")
print(output_path)


# ============================================================
# 11. DOWNLOAD RESULT
# ============================================================

files.download(output_path)


# ============================================================
# 12. RELEASE
# ============================================================

face_mesh.close()
```

```python
import base64
import cv2
import numpy as np
import time
from IPython.display import display, HTML
from google.colab.output import eval_js

# 1. Inject HTML and JavaScript interface
display(HTML('''
<div style="font-family: sans-serif; padding: 10px;">
  <p id="cam-status" style="font-weight: bold; color: orange;">Requesting camera access...</p>
  <div style="display: flex; gap: 10px;">
    <!-- Raw live preview so you immediately know the camera is working -->
    <div>
      <small>Webcam Direct Feed</small><br>
      <video id="raw-video" width="400" height="300" autoplay playsinline muted style="border: 1px solid #ccc;"></video>
    </div>
    <!-- Processed canvas feed returned from OpenCV -->
    <div>
      <small>OpenCV Drawing Canvas</small><br>
      <img id="processed-stream" width="400" height="300" style="border: 1px solid #000; background: #000;" />
    </div>
  </div>
  <br>
  <button id="btn-stop" style="padding: 8px 16px; background: #e74c3c; color: white; border: none; border-radius: 4px; cursor: pointer;">Stop Tracking</button>
  <button id="btn-clear" style="padding: 8px 16px; background: #3498db; color: white; border: none; border-radius: 4px; cursor: pointer;">Clear Canvas</button>
</div>

<script>
window.isColabActive = true;
window.clearCanvasRequested = false;
let videoElem = document.getElementById('raw-video');
let canvasElem = document.createElement('canvas');
canvasElem.width = 640;
canvasElem.height = 480;
let ctx = canvasElem.getContext('2d');

document.getElementById('btn-stop').onclick = () => {
    window.isColabActive = false;
    document.getElementById('cam-status').innerText = 'Stopped.';
    document.getElementById('cam-status').style.color = 'red';
};

document.getElementById('btn-clear').onclick = () => {
    window.clearCanvasRequested = true;
};

// Initialize Camera
async function startWebcam() {
    try {
        const stream = await navigator.mediaDevices.getUserMedia({
            video: { width: 640, height: 480, facingMode: 'user' }
        });
        videoElem.srcObject = stream;
        await videoElem.play();
        window.cameraReady = true;
        document.getElementById('cam-status').innerText = 'Camera connected. Tracking active!';
        document.getElementById('cam-status').style.color = 'green';
    } catch (err) {
        document.getElementById('cam-status').innerText = 'Camera access error: ' + err.name + ' - ' + err.message;
        document.getElementById('cam-status').style.color = 'red';
        window.cameraReady = false;
    }
}

window.captureFrame = () => {
    ctx.drawImage(videoElem, 0, 0, 640, 480);
    return canvasElem.toDataURL('image/jpeg', 0.7);
};

window.renderOutput = (base64Img) => {
    document.getElementById('processed-stream').src = base64Img;
};

startWebcam();
</script>
'''))

# 2. Wait until browser grants camera permission
print("Waiting for browser camera permission...")
while True:
    try:
        ready = eval_js('window.cameraReady')
        if ready is True:
            print("Webcam successfully connected.")
            break
        elif ready is False:
            raise RuntimeError("Webcam access was denied or failed in browser.")
    except Exception:
        pass
    time.sleep(0.5)

# 3. Helpers to convert images
def js_to_cv(js_data):
    header, encoded = js_data.split(',', 1)
    img_data = base64.b64decode(encoded)
    np_arr = np.frombuffer(img_data, dtype=np.uint8)
    return cv2.imdecode(np_arr, cv2.IMREAD_COLOR)

def cv_to_js(cv_img):
    _, buffer = cv2.imencode('.jpg', cv_img, [cv2.IMWRITE_JPEG_QUALITY, 70])
    b64 = base64.b64encode(buffer).decode('utf-8')
    return f"data:image/jpeg;base64,{b64}"

# 4. Canvas setup & tracking loop
canvas_draw = np.zeros((480, 640, 3), dtype=np.uint8)
prev_point = None

# Target color HSV range (defaults to bright Blue)
COLOR_LOWER = np.array([90, 80, 80])
COLOR_UPPER = np.array([135, 255, 255])

print("Hold a blue object (marker, pen, bottle cap) to draw.")

while True:
    active = eval_js('window.isColabActive')
    if not active:
        break

    # Check if clear button was pressed
    if eval_js('window.clearCanvasRequested'):
        canvas_draw[:] = 0
        eval_js('window.clearCanvasRequested = false')

    # Read current webcam frame
    frame_data = eval_js('window.captureFrame()')
    if not frame_data:
        continue

    frame = js_to_cv(frame_data)
    frame = cv2.flip(frame, 1)  # Mirror view

    # Detect object
    blurred = cv2.GaussianBlur(frame, (11, 11), 0)
    hsv = cv2.cvtColor(blurred, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv, COLOR_LOWER, COLOR_UPPER)
    mask = cv2.erode(mask, None, iterations=2)
    mask = cv2.dilate(mask, None, iterations=2)

    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    curr_point = None

    if contours:
        c = max(contours, key=cv2.contourArea)
        if cv2.contourArea(c) > 600:
            ((x, y), radius) = cv2.minEnclosingCircle(c)
            M = cv2.moments(c)
            if M["m00"] > 0:
                curr_point = (int(M["m10"] / M["m00"]), int(M["m01"] / M["m00"]))
                cv2.circle(frame, (int(x), int(y)), int(radius), (0, 255, 255), 2)

    # Draw lines
    if prev_point and curr_point:
        cv2.line(canvas_draw, prev_point, curr_point, (0, 255, 0), 5)
    prev_point = curr_point

    # Merge frame with canvas
    canvas_gray = cv2.cvtColor(canvas_draw, cv2.COLOR_BGR2GRAY)
    _, mask_inv = cv2.threshold(canvas_gray, 10, 255, cv2.THRESH_BINARY_INV)
    frame_bg = cv2.bitwise_and(frame, frame, mask=mask_inv)
    output = cv2.add(frame_bg, canvas_draw)

    # Push processed image back to browser
    eval_js(f'window.renderOutput("{cv_to_js(output)}")')
```

```python
# ============================================================
# SPAM DETECTION USING MACHINE LEARNING
# Naive Bayes + TF-IDF
# Google Colab
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report


# ============================================================
# 2. CREATE DATASET
# ============================================================

data = {
    "Message": [
        "Congratulations! You won a free lottery",
        "Win a free iPhone now",
        "Claim your cash prize immediately",
        "You have won 10000 dollars",
        "Click here to get your free reward",
        "Congratulations you are the lucky winner",
        "Free entry in a weekly competition",
        "You have won a free vacation",
        "Get free money now",
        "Urgent! Claim your prize",

        "Hey, are we meeting today?",
        "Please send me the assignment",
        "Can you call me when you are free?",
        "Your appointment is tomorrow",
        "Let's meet for lunch",
        "The project meeting starts at 10 AM",
        "Please submit your assignment today",
        "Happy birthday! Have a great day",
        "Can you send me the notes?",
        "I will reach home by 7 PM",

        "You won a cash prize, click now",
        "Free vacation waiting for you",
        "Claim your lottery money",
        "Congratulations! You won a prize",
        "Get your free gift today",

        "Please call me after class",
        "I am going to college now",
        "Can you share the project file?",
        "Meeting is scheduled for tomorrow",
        "Thank you for your help"
    ],

    "Label": [
        "spam","spam","spam","spam","spam",
        "spam","spam","spam","spam","spam",

        "ham","ham","ham","ham","ham",
        "ham","ham","ham","ham","ham",

        "spam","spam","spam","spam","spam",

        "ham","ham","ham","ham","ham"
    ]
}

df = pd.DataFrame(data)


# ============================================================
# 3. DISPLAY DATASET
# ============================================================

print("Dataset:")
print(df)

print("\nDataset Shape:")
print(df.shape)


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# 5. DATA CLEANING
# ============================================================

df["Message"] = df["Message"].str.lower()

df["Message"] = df["Message"].fillna("")


# ============================================================
# 6. INPUT AND TARGET
# ============================================================

X = df["Message"]
y = df["Label"]


# ============================================================
# 7. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))


# ============================================================
# 8. TEXT TO NUMERICAL FEATURES
# ============================================================

vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)

X_test_tfidf = vectorizer.transform(X_test)


print("\nTF-IDF Training Shape:")
print(X_train_tfidf.shape)


# ============================================================
# 9. CREATE NAIVE BAYES MODEL
# ============================================================

model = MultinomialNB()


# ============================================================
# 10. TRAIN MODEL
# ============================================================

model.fit(
    X_train_tfidf,
    y_train
)

print("\nModel Training Completed")


# ============================================================
# 11. MAKE PREDICTION
# ============================================================

y_pred = model.predict(X_test_tfidf)

print("\nActual Values:")
print(y_test.values)

print("\nPredicted Values:")
print(y_pred)


# ============================================================
# 12. ACCURACY
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nAccuracy:",
      accuracy * 100,
      "%")


# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


# ============================================================
# 14. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)


# ============================================================
# 15. TEST NEW MESSAGES
# ============================================================

new_messages = [
    "Congratulations you won a free prize",
    "Can you send me the project notes?",
    "Claim your free cash reward now",
    "Are you coming to college tomorrow?"
]


# Convert new messages into TF-IDF
new_messages_tfidf = vectorizer.transform(
    new_messages
)


# Predict
predictions = model.predict(
    new_messages_tfidf
)


# ============================================================
# 16. DISPLAY PREDICTIONS
# ============================================================

print("\nNew Message Predictions:")

for message, prediction in zip(
    new_messages,
    predictions
):

    print("\nMessage:", message)
    print("Prediction:", prediction.upper())


# ============================================================
# 17. PREDICTION PROBABILITY
# ============================================================

probabilities = model.predict_proba(
    new_messages_tfidf
)

print("\nPrediction Probabilities:")

for message, probability in zip(
    new_messages,
    probabilities
):

    print("\nMessage:", message)

    print(
        "Ham:",
        round(probability[0] * 100, 2),
        "%"
    )

    print(
        "Spam:",
        round(probability[1] * 100, 2),
        "%"
    )
```
