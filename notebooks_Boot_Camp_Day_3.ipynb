```python
# STUDENT PERFORMANCE - MACHINE LEARNING

# 1. Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# 2. LOAD DATA
df = pd.read_csv("student_performance.csv")

print("Original Data:")
print(df.head())

print("\nShape:")
print(df.shape)
```

```python
# 3. DATA CLEANING

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing numerical values with column mean
df["Study_Hours"] = df["Study_Hours"].fillna(
    df["Study_Hours"].mean()
)

df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].mean()
)

df["Previous_Score"] = df["Previous_Score"].fillna(
    df["Previous_Score"].mean()
)

df["Assignments_Completed"] = df["Assignments_Completed"].fillna(
    df["Assignments_Completed"].mean()
)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# 4. INPUT AND TARGET

# Input features
X = df[
    [
        "Study_Hours",
        "Attendance",
        "Previous_Score",
        "Assignments_Completed"
    ]
]

# Target variable
y = df["Final_Score"]

print("\nInput X:")
print(X)

print("\nTarget y:")
print(y)
```

```python
# 5. RESHAPE

# Convert X and y into NumPy arrays
X = np.array(X)
y = np.array(y)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
# 6. TRAIN-TEST SPLIT


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Input Shape:", X_train.shape)
print("Testing Input Shape:", X_test.shape)

print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

```

```python
# 7. MODEL IMPLEMENTATION

model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("\nModel Training Completed")
```

```python
# 8. MAKE PREDICTION


y_pred = model.predict(X_test)

print("\nActual Values:")
print(y_test)

print("\nPredicted Values:")
print(y_pred)
```

```python
# 9. PREDICT FOR A NEW STUDENT

# New student:
# Study Hours = 6
# Attendance = 90
# Previous Score = 75
# Assignments Completed = 9

new_student = np.array([
    [6, 90, 75, 9]
])

prediction = model.predict(new_student)

print("\nNew Student Predicted Final Score:")
print(prediction[0])
```

```python
# 10. MODEL EVALUATION

# Calculate prediction error
error = np.abs(y_test - y_pred)

# Accuracy for each student
accuracy = 100 - (error / y_test * 100)

# Average accuracy
average_accuracy = np.mean(accuracy)

print("Accuracy for each student:")
print(accuracy)

print("\nAverage Accuracy:", average_accuracy, "%")

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nMean Squared Error:", mse)
print("R2 Score:", r2)
```

```python
# ==========================================
# DECISION TREE - MACHINE FAILURE PREDICTION
# ==========================================

# 1. Import libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report


```

```python
# ==========================================
# 2. LOAD DATASET
# ==========================================

df = pd.read_csv("factory_machine_failure_prediction.csv")

print("Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)


```

```python

# ==========================================
# 3. CHECK DATA
# ==========================================

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
print(df.info())
```

```python

# ==========================================
# 4. DATA CLEANING
# ==========================================

# Fill missing numerical values with mean

numeric_columns = [
    "Temperature_C",
    "Vibration_mm_s",
    "Pressure_psi",
    "RPM",
    "Operating_Hours"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())


print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python

# ==========================================
# 5. INPUT AND TARGET
# ==========================================

X = df[
    [
        "Temperature_C",
        "Vibration_mm_s",
        "Pressure_psi",
        "RPM",
        "Operating_Hours"
    ]
]

y = df["Machine_Failure"]


print("\nInput X:")
print(X.head())

print("\nTarget y:")
print(y.head())
```

```python
# ==========================================
# 6. CONVERT TO NUMPY
# ==========================================

X = np.array(X)
y = np.array(y)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
```

```python


# ==========================================
# 7. TRAIN-TEST SPLIT
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
# ==========================================
# 8. CREATE DECISION TREE MODEL
# ==========================================

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=4,
    random_state=42
)
```

```python

# ==========================================
# 9. TRAIN MODEL
# ==========================================

model.fit(X_train, y_train)

print("\nDecision Tree Model Training Completed")
```

```python
# ==========================================
# 10. MAKE PREDICTION
# ==========================================

y_pred = model.predict(X_test)

print("\nActual Values:")
print(y_test)

print("\nPredicted Values:")
print(y_pred)
```

```python

# ==========================================
# 11. ACCURACY
# ==========================================

accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy * 100, "%")
```

```python


# ==========================================
# 12. CONFUSION MATRIX
# ==========================================

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)
```

```python

# ==========================================
# 13. CLASSIFICATION REPORT
# ==========================================

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# ==========================================
# 14. PREDICT A NEW MACHINE
# ==========================================

# New machine:
# Temperature = 90 C
# Vibration = 7.2 mm/s
# Pressure = 80 psi
# RPM = 1950
# Operating Hours = 4500

new_machine = np.array([
    [90, 7.2, 80, 1950, 4500]
])

prediction = model.predict(new_machine)

print("\nNew Machine Prediction:")

if prediction[0] == 1:
    print("Machine Failure: YES")
else:
    print("Machine Failure: NO")
```

```python

# ==========================================
# 15. FAILURE PROBABILITY
# ==========================================

probability = model.predict_proba(new_machine)

print("\nFailure Probability:",
      probability[0][1] * 100, "%")
```
