```python
#list
my_list = ["Apple", "Banana", "Mango", "Orange"]

for i in range(len(my_list)):
    print(my_list[i])
```

```python
#string split usign for loop
text = "Python is easy to learn"

words = text.split()

for i in range(len(words)):
    print(words[i])
```

```python
#access the element by index
text = "Python Java SQL"

words = text.split()

print(words[0])
print(words[1])
print(words[2])
```

```python
#for loop to print elements in list
fruits = ["Apple", "Banana", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
```

```python
#function
def print_fruits():
    fruits = ["Apple", "Banana", "Mango", "Orange"]

    for fruit in fruits:
        print(fruit)
```

```python
#function call
print_fruits()
```
